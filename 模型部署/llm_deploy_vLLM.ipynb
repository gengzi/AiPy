## 使用 vLLM 框架部署模型

### 简介
**vLLM框架**是一个常用的部署框架，它的优势如下：
1. 更快的推理速度
2. 处理并发请求
3. 接口简洁，封装程度高

缺点: 占用的显存的会比较大，vllm本身会占有大量的显存

### 进行对话

```python
# 导入 vllm库中的两个类，LLM是用于加载模型的类，SamplingParams是用于封装模型调用参数的类
from vllm import LLM, SamplingParams

# 使用 vllm 加载模型
model_path = "model/DeepSeek-R1-Distill-Qwen-1.5B"
llm = LLM(model=model_path)

# 定义采样参数
sampling_params = SamplingParams(temperature=0.7, top_p=0.8, max_tokens=512, repetition_penalty=1.1)
```

```python
# 定义用户输入
prompt = "1+2等于几"

# 进行推理
outputs = llm.generate(prompt, sampling_params)

# 输出结果
print(outputs[0].outputs[0].text)
```

### 多轮对话

```python
history = []
query = input("请输入内容：")
while query:
    # 构建多轮对话的提示
    messages = [{"role": "system", "content": "你是一个智能助手，擅长用中文回答用户的提问。"}]
    for query_h, response_h in history:
        messages.append({"role": "user", "content": query_h})
        messages.append({"role": "assistant", "content": response_h})
    messages.append({"role": "user", "content": query})

    # 直接使用 messages作为输入，调用 chat方法来生成回复
    outputs = llm.chat(messages, sampling_params)
    response = outputs[0].outputs[0].text

    print(response)
    print("-" * 100)

    # 将本轮对话的历史加入 history列表中
    history.append([query, response])
    query = input("请输入内容：")
```
