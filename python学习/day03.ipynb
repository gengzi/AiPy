```python


# 模拟生成一个csv文件
# 定义一个列表字典就是map
data = [
    ["question", "answer"],
    ["解释一下JVM内存模型", "JVM内存模型主要包括堆、栈、方法区..."],
    ["什么是Python的GIL?", "GIL是全局解释器锁，它限制了多线程并行..."]
]

import csv

def create_csv():
    """
    模拟生成一个csv文件
    :return: csv文件
    """
    with open("/mnt/f/baidu/testdata.csv", "w",encoding='utf-8',newline="") as f:
        writer = csv.writer(f)
        writer.writerows(data)
    print(f"✅文件写入成功")


json = {
  "instruction": "请分析这段 Java 代码的潜在死锁风险。",
  "input": "可以不填充",
  "output": "这段代码存在死锁风险，因为...（详细的分析过程）..."
}

import json

def csv_to_jsonl(csvfilePath:str):
    """
    csv文件转jsonl文件
    :param csvfilePath: csv文件路径
    :return: jsonl文件
    """
    # 定义一个列表
    jsonl_data =  []

    with open(csvfilePath, "r",encoding='utf-8') as f:
        reader = csv.DictReader(f)
        # 获取每一行数据
        for row in reader:
            jsonl_data.append({
                "instruction": row["question"],
                "input": "",
                "output": row["answer"]
            })

    with open("/mnt/f/baidu/testdata.jsonl", "w",encoding='utf-8') as f:
        for data in jsonl_data:
            json.dump(data, f, ensure_ascii=False)
            f.write("\n")

    print(f"写入完成：{jsonl_data}")




if __name__ == "__main__":
    create_csv()
    csv_to_jsonl("/mnt/f/baidu/testdata.csv")









```

```python
def csv_to_jsonl(csvfilePath: str):
    """
    csv文件转jsonl文件（边读边写）
    :param csvfilePath: csv文件路径
    :return: jsonl文件
    """
    with open(csvfilePath, "r", encoding='utf-8') as infile, \
         open("/mnt/f/baidu/testdata.jsonl", "w", encoding='utf-8') as outfile:

        reader = csv.DictReader(infile)

        # 边读边写
        for row in reader:
            json_line = {
                "instruction": row["question"],
                "input": "",
                "output": row["answer"]
            }
            json.dump(json_line, outfile, ensure_ascii=False)
            outfile.write("\n")

    print("✅ 文件写入成功")
```
