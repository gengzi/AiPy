```python
import os


fileType = {
    "图片": [".jpg", ".jpeg", ".png", ".gif", ".bmp"],
    "文档": [".pdf", ".docx", ".txt", ".xlsx", ".pptx"],
    "安装包": [".exe", ".msi", ".dmg"],
    "压缩文件": [".zip", ".rar", ".7z"]
}


## 定义一个方法
def organize_file(path: str):
    """
    将文件进行整理
    """
    if not os.path.exists(path):
        print("不存在该路径")
        return

    # 循环文件列表
    for fileName in os.listdir(path):
        print(fileName)
        pfileName = os.path.join(path,fileName)
        # 判断是否为文件夹
        if(os.path.isdir(pfileName)):
             continue
        # 将文件后缀名小写
        _,ext = os.path.splitext(fileName)
        ext = ext.lower()

        for key,value in fileType.items():
            print(f"key:{key},value:{value}")
            if ext in value:
                print(f"存在该类型的数据：{ext}")
            else:
                print(f"不存在该类型数据：{ext}")







if __name__ == "__main__":
    path = r"/mnt/f/baidu"
    organize_file(path)
```
