```python
import uvicorn
import nest_asyncio
from fastapi import FastAPI
from pydantic import BaseModel
from typing import List

# 1. 应用补丁：允许在 Jupyter 的循环中嵌套运行 Uvicorn
nest_asyncio.apply()
app = FastAPI()


class Request(BaseModel):
    name:str
    age:int
    city:List[str]

class Response(BaseModel):
    code:int
    message:str
    data:dict

@app.post("/api")
def api(request:Request):
    """
    api接口
    :param request:
    :return: 保存成功
    """
    response =  Response(code=200,message="保存成功",data={"name":request.name})
    return response


# 2. 启动服务器
# 注意：这行代码运行后，当前 Cell 会一直显示 [*] (运行中状态)
# 这是正常的，因为它正在监听 HTTP 请求。
if __name__ == "__main__":
    config = uvicorn.Config(app, host="127.0.0.1", port=8000, log_level="info")
    server = uvicorn.Server(config)
    # 使用 asyncio.run() 启动服务器
    import asyncio
    asyncio.run(server.serve())












```
